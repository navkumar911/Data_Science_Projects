To accurately answer our core business questions—identifying top revenue-driving products and segmenting our most valuable customers—our dataset must reflect only successful, completed sales. During exploratory data analysis, we identified three major anomalies that actively distort our revenue and customer metrics:

1. Negative Unit Prices (Accounting Adjustments): Values like -11,062.06 are labeled as "Adjust bad debt." These are internal accounting tricks to balance the books when a customer defaults on payment, not real product sales. Keeping them would artificially subtract thousands of dollars from our total revenue.

2. Extreme Negative Quantities (Inventory Mismanagement): Massive negative quantities (e.g., -80,995 or -9,600) frequently correspond to a price of 0.00 and descriptions like "thrown away" or "printing smudges." These are warehouse workers writing off damaged inventory, completely unrelated to customer buying behavior.

3. Normal Negative Quantities (Cancellations & Returns): Rows starting with 'C' represent canceled orders. Since our objective is to calculate actual revenue generated and segment customers based on what they bought and kept, including returns will skew our Recency, Frequency, and Monetary (RFM) modeling.

Action Taken: We are filtering the dataset to strictly include rows where both Quantity > 0 and UnitPrice > 0. This isolates pure, successful transactions for accurate modeling.

# RFM Analysis

```python
import datetime as dt
```

```python
max_date = df["InvoiceDate"].max()
print(max_date)
```

```python
snapshot_date = max_date + dt.timedelta(days=1)
print(snapshot_date)
```

To calculate Recency (how many days it has been since a customer last bought something), we need a reference point for "today." Since this dataset is from 2011, we can't use the actual current date, or every customer will look like they haven't bought anything in 15 years!

Instead, we find the very last transaction in the entire dataset and set our "Snapshot Date" to exactly one day after that.

```python
# Filtering out anonymous transactions
rfm_df = df[df["CustomerID"] != "Unknown"].copy()
```

```python
rfm = rfm_df.groupby("CustomerID").agg({
    "InvoiceDate" : lambda x: (snapshot_date - x.max()).days,  # Recency: Days since last purchase
    "InvoiceNo" : "nunique",                                   # Frequency: Count of unique invoices
    "Revenue" : "sum"                                          # Monetary: Total Money Spent
})

# Renaming the columns to standard RFM naming
rfm.rename(columns={
    "InvoiceDate" : "Recency",
    "InvoiceNo" : "Frequency",
    "Revenue" : "Monetary"
}, inplace = True)

rfm.head()
```

```python
# Creating a figure with 3 subplots
fig, axes = plt.subplots(1,3, figsize = (18,5))

# Plotting recency
sns.boxplot(x=rfm["Recency"], ax = axes[0])
axes[0].set_title("Distribution of Recency (Days)")

# Plotting Frequency
sns.boxplot(x=rfm["Frequency"], ax = axes[1])
axes[1].set_title("Distribution of Frequency (Invoices)")

# Plotting Monetary
sns.boxplot(x=rfm["Monetary"], ax = axes[2])
axes[2].set_title("Distribution of Monetary (Total and Spent)")

plt.tight_layout()
plt.show()
```

```python
whales = rfm.sort_values(by="Monetary", ascending=False).head(10)
print(whales)
```

RFM Outlier Strategy: Extracting the VIPs
1. Observation:
An inspection of the top monetary customers reveals they are not data errors or one-time anomalies. The majority have extremely high frequency (e.g., 50 to 199 invoices) and very low recency, indicating they are highly engaged, loyal B2B wholesale partners.

2. Action Taken (VIP Extraction):
We are extracting these extreme "Whales" into a dedicated VIP list rather than keeping them in the main modeling dataset.

3. Machine Learning Justification:
K-Means clustering relies on Euclidean distance and is highly sensitive to extreme variance. If we include customers spending £270,000+ alongside average retail customers spending £300, the algorithm's centroids will be violently skewed. The model will create isolated micro-clusters for individual VIPs while failing to meaningfully segment the remaining 99% of the normal customer base. Separating them allows for dedicated VIP account management while protecting the mathematical integrity of the algorithm.

```python
vip_customers = rfm[rfm["Monetary"] > 50000]

rfm_clean = rfm[rfm["Monetary"] <= 50000].copy()

print(f"Number of VIPs extracted: {len(vip_customers)}")
print(f"Number of regular customers for K-Means: {len(rfm_clean)}")
```

RFM Outlier Strategy: Capping (Winsorization)
The Challenge (The "Mini-Whales"):
Although we extracted the extreme £50,000+ VIPs, our data still contains "mini-whales" (e.g., customers who spent £30,000 or ordered 150 times). We do not want to remove these high-value customers from our segmentation model, but leaving their extreme raw numbers untouched will skew the K-Means algorithm.

Capping vs. Scaling:
To prepare this data for machine learning, we must perform two distinct steps:

1. Capping (Fixing the Outliers): We set a mathematical "ceiling" on the data. If a customer's spending exceeds the 99th percentile, we do not delete them; we simply reduce their value to match that exact 99th percentile limit. This neutralizes their gravitational pull on the algorithm while keeping them in the high-value segment.

2. Scaling (Fixing the Units): K-Means cannot naturally compare Days (Recency) to Dollars (Monetary). After capping, we will scale the data so all three metrics share the exact same mathematical weight.

Crucially, capping must happen BEFORE scaling. If we scale first, the extreme outliers will compress the normal customers together, destroying the algorithm's ability to segment average shoppers.

```python
q_freq = rfm_clean["Frequency"].quantile(0.99)
q_money = rfm_clean["Monetary"].quantile(0.99)

print(f"99th Percentile Frequency Cap: {q_freq} invoices")
print(f"99th Percentile Monetary Cap: £{q_money:.2f}")

rfm_clean["Frequency"] = rfm_clean["Frequency"].clip(upper=q_freq)
rfm_clean["Monetary"] = rfm_clean["Monetary"].clip(upper=q_money)

print("\nMax values after capping:")
print(rfm_clean.max())
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()

rfm_scaled_array = scaler.fit_transform(rfm_clean)

rfm_scaled = pd.DataFrame(rfm_scaled_array, columns = rfm_clean.columns, index = rfm_clean.index)

display(rfm_scaled.head())
```

Phase 2 Complete: Interpreting the Scaled Data
How to Read the Scaled RFM Table:
By applying the StandardScaler, we have translated Days, Counts, and Pounds into a single universal metric based on standard deviations from the mean.

1. 0.0 represents the exact average for that metric across all normal customers.

2. Positive numbers mean the customer is above average.

3. Negative numbers mean the customer is below average.

Reading a Customer Profile at a Glance:
Looking directly at Customer 12347.0 in our scaled table:

1. Recency (-0.90): Below average days since last purchase. (Positive business signal: They shopped recently).

2. Frequency (0.71): Above average number of orders. (Positive business signal: They are frequent buyers).

3. Monetary (1.33): They spend more than 1 full standard deviation above the average customer. (Positive business signal: High spender).

Without looking at a single raw dollar amount, the algorithm can now easily identify this as a highly valuable customer. This mathematically balanced dataset is perfectly optimized for clustering.

# Phase 3: Unsupervised Machine Learning (K-Means)

Next Step: The Elbow Method

Before we can run the algorithm, we must tell K-Means how many clusters (customer segments) to create. If we guess blindly, we could ruin the segmentation. We will run the model 10 separate times (K=1 through 10) and plot the error rate (WCSS). We are looking for the "elbow" of the graph—the exact point where the error drops sharply and then flattens out. That point dictates our optimal number of customer segments.

```python
from sklearn.cluster import KMeans
```

```python
wcss = []

for i in range(1, 11):
    # # Initialize the model (random_state ensures we get the same results every time)
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)

    kmeans.fit(rfm_scaled)

    # Grab the WCSS (inertia) and add it to our list
    wcss.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(10,6))
plt.plot(range(1,11), wcss, marker="o", linestyle = "--", color="b")
plt.title("The Elbow Method (Finding the optimal K)")
plt.xlabel("Number of clusters (K)")
plt.ylabel("WCSS (Error/Inertia)")
plt.grid(True)
plt.show()
```

Phase 3: The Elbow Method Results & K-Means Execution

Interpreting the Elbow Curve:

The Within-Cluster Sum of Squares (WCSS) drops precipitously from K=1 to K=3. At K=3, the rate of variance reduction slows significantly, forming a distinct "elbow" in the graph. While K=4 also offers a reduction in error, the most mathematically significant bend occurs at K=3.

Execution Strategy (3-Tier Segmentation):
We will proceed with K=3, which perfectly aligns with a highly actionable 3-tier marketing segmentation strategy:

1. Cluster A: Loyal & High-Value Customers (Keep & Grow)

2. Cluster B: Mid-Value Customers (Nurture & Convert)

3. Cluster C: At-risk / Inactive Customers (Win-back or Let Go)

We will run the final K-Means model on our scaled dataset (rfm_scaled). Once the algorithm assigns a cluster number to each customer, we will attach those labels back to our unscaled dataset (rfm_clean) so we can profile the clusters using actual business units (Days, Counts, and Pounds).

```python
kmeans_final = KMeans(n_clusters=3, init='k-means++', random_state=42, n_init=10)

cluster_labels = kmeans_final.fit_predict(rfm_scaled)

rfm_clean["Cluster"] = cluster_labels

display(rfm_clean.head())
```

```python
# Calculating the average RFM values and the number of customers for each cluster
cluster_summary = rfm_clean.groupby("Cluster").agg({
    "Recency" : "mean",
    "Frequency" : "mean",
    "Monetary" : ["mean", "count"] # 'count' tells us how many people are in the cluster
}).round(1)

# Cleaning up the columns for readbility
cluster_summary.columns = ["Avg_recency (Days)", "Avg_Frequency (Orders)", "Avg_Monetary (£)", "Customer_Count"]

# Display the profile of our clusters
display(cluster_summary)
```

Interpreting the K-Means Clusters:

By calculating the mean Recency, Frequency, and Monetary values for our three machine-learned segments, we can clearly define the real-world identity of each group:

1. Cluster 1 (High-Value / Champions):

Profile: High spenders (£7,208 avg), very frequent buyers (15.6 orders), and highly active (last shopped 19 days ago).

3. Cluster 0 (At-Risk / Lost Customers):

```python
# 1. Mapping the cluster numbers to our human-readable personas
segment_names = {
    0: "At-Risk/ Lost",
    1: "Champions",
    2: "Core Regulars"
}

# 2. Creating the new Segment column in the rfm_clean
rfm_clean["Segment"] = rfm_clean["Cluster"].map(segment_names)

# 3. Checking the final table to confirm the new column exists
display(rfm_clean[["Recency", "Frequency", "Monetary", "Cluster", "Segment"]].head())
```
